# READ `GEOLOCATIONS` FROM VOLUME

```python
# Define GeoLocation Table Schema with Data Types
geo_schema = '''geolocation_zip_code_prefix INT,
geolocation_lat DOUBLE,
geolocation_lng DOUBLE,
geolocation_city STRING,
geolocation_state STRING'''
```

```python
# Load GeoLocation Data from CSV Files (reads both files in the folder)
geo_df = spark.read.format('csv') \
    .option('header', True) \
    .schema(geo_schema) \
    .load('/Volumes/sales/landing/sales/GeoLocations/')
```

```python
# Display GeoLocation Data and Schema
print(f"Rows: {geo_df.count()}")
display(geo_df.limit(3))
geo_df.printSchema()
```

```python
# Write GeoLocations to Bronze Delta Table
geo_df.write.format('delta').mode('overwrite').saveAsTable('sales.bronze.geolocations')
```
