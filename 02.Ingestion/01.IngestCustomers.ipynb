# READ `CUSTOMERS` FROM VOLUME

```python
# Define Customer Table Schema with Data Types
cust_schema = '''customer_id STRING,
customer_unique_id STRING,
customer_zip_code_prefix INT,
customer_city STRING,
customer_state STRING'''
```

```python
# Load Customer Data from CSV Files (reads all files in the folder)
cust_df = spark.read.format('csv') \
    .option('header', True) \
    .schema(cust_schema) \
    .load('/Volumes/sales/landing/sales/Customers/')
```

```python
# Display Customer Data and Schema
print(f"Rows: {cust_df.count()}")
display(cust_df.limit(3))
cust_df.printSchema()
```

```python
# Write Customers to Bronze Delta Table
cust_df.write.format('delta').mode('overwrite').saveAsTable('sales.bronze.customers')
```
