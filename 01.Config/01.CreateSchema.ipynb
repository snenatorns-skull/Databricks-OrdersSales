# CREATE SCHEMAS AND LANDING VOLUME

```python
%sql
-- Create the medallion schemas in the sales catalog
CREATE SCHEMA IF NOT EXISTS sales.landing;
CREATE SCHEMA IF NOT EXISTS sales.bronze;
CREATE SCHEMA IF NOT EXISTS sales.silver;
CREATE SCHEMA IF NOT EXISTS sales.gold;
```

```python
%sql
-- Create the landing volume that holds the raw CSV files
CREATE VOLUME IF NOT EXISTS sales.landing.sales;
```

```python
# Create one landing folder per source table and list its files
base_path = "/Volumes/sales/landing/sales"
folders = ["Customers", "GeoLocations", "OrderItems", "OrderPayments", "OrderReviews", "Orders", "Product_Category_Translation", "Products", "Sellers"]

for folder in folders:
    dbutils.fs.mkdirs(f"{base_path}/{folder}")
    files = dbutils.fs.ls(f"{base_path}/{folder}")
    print(f"{folder}: {[f.name for f in files]}")
```
